```python
# Import Librares

import pandas as pd
import os
```

```python
# Load Datasets

orders = pd.read_csv("../data/raw/orders.csv")
products = pd.read_csv("../data/raw/products.csv")
marketing = pd.read_csv("../data/raw/marketing_spend.csv")
```

# Data type transformation

## Convert order_date from orders

```python
orders["order_date"].dtype
```

```python
orders["order_date"] = pd.to_datetime(orders["order_date"])
```

## Convert month from marketing

```python
marketing["month"].dtype
```

```python
marketing["month"] = pd.to_datetime(marketing["month"])
```

## Verify all data types after transformation

```python
orders.info()
```

```python
marketing.info()
```

# Validate Cleaned Datasets

## Orders

```python
orders.isnull().sum()
```

```python
orders["order_id"].duplicated().sum()
```

```python
# Checking profit calculation

(orders["net_revenue"] - orders["total_costs"] - orders["profit"]).sum()
```

```python
# Checking total_cost calculation

(
orders["product_cost"] +
orders["shipping_cost"] +
orders["platform_fee"] +
orders["transaction_fee"]
- orders["total_costs"]
).sum()
```

## Marketing

```python
marketing.isnull().sum()
```

```python
(
    marketing["spend"] / marketing["conversions"]
    - marketing["cpa"]
).abs().max()
```

```python
(
    marketing["spend"] / marketing["clicks"]
    - marketing["cpc"]
).abs().max()
```

```python
(
    marketing["revenue_attributed"] / marketing["spend"]
    - marketing["roas"]
).abs().max()
```
